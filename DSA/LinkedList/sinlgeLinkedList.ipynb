### Linked List 
Linked list is a linear data structure . and not continous store data in nodes .

```python
class Node:
    def __init__(self,data=None,next=None):
        self.data = data
        self.next = next
class LinkedList:
    def __init__(self,head=None):
        self.head=head
        self.n=0
#Len........................................
    def __len__(self):
        return self.n

#ISempty..........................................
    def isempty(self):
        return self.head==None

#Insert at first......................................................................
    def insertatfirst(self,data):
        new_node=Node(data,self.head)
        self.head=new_node
        self.n=self.n+1

#Insert at last......................................................................
    def insertatlast(self,data):
        new_node=Node(data)
        if not self.isempty():
            temp=self.head
            while temp.next is not None:
                temp=temp.next
            temp.next=new_node
        else:
            self.head=new_node
        self.n=self.n+1

# Insert at mid................................................
    def insert_at_mid(self,data,pos):
        new_node=Node(data)
        step=0
        temp=self.head
        while temp.next is not None:
            step=step+1
            if step==pos:
                new_node.next=temp.next
                temp.next=new_node
                self.n=self.n+1
                return
            temp=temp.next
        print("Your given pos is out of range")

    
    def printL(self):
        temp=self.head
        while temp is not None:
            print(temp.data)
            temp=temp.next
        pass


l=LinkedList()
l.insertatfirst(45)
l.insertatfirst(50)
l.insertatfirst(12)
l.insertatlast(1)
l.insertatlast(2)
l.insert_at_mid(4,1)
l.printL()
print("lenght:",len(l))
```
