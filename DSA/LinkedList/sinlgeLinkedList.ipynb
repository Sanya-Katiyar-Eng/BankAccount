```python
class Node:
    def __init__(self,data=None,next=None):
        self.data = data
        self.next = next
class LinkedList:
    def __init__(self,head=None):
        self.head=head
        self.n=0
#Len........................................
    def __len__(self):
        return self.n

#ISempty..........................................
    def isempty(self):
        return self.head==None

#Insert at first......................................................................
    def insertatfirst(self,data):
        new_node=Node(data,self.head)
        self.head=new_node
        self.n=self.n+1

#Insert at last......................................................................
    def insertatlast(self,data):
        new_node=Node(data)
        if not self.isempty():
            temp=self.head
            while temp.next is not None:
                temp=temp.next
            temp.next=new_node
        else:
            self.head=new_node
        self.n=self.n+1

# Insert at mid................................................
    def insert_at_mid(self,data,pos):
        new_node=Node(data)
        step=0
        temp=self.head
        while temp.next is not None:
            step=step+1
            if step==pos:
                new_node.next=temp.next
                temp.next=new_node
                self.n=self.n+1
                return
            temp=temp.next
        print("Your given pos is out of range")
#delete at first................................................
    def delete_at_first(self):

        if not self.isempty():
            self.head=self.head.next
            self.n=self.n-1
            return
        print("Your Linked List has no element ")
#delete at last..................................................
    def delete_at_last(self):
        if not self.isempty():
            temp=self.head
            while temp.next.next is not None:
                temp=temp.next
            temp.next=None
            self.n=self.n-1
            return
        print("Your linked list empty")
#delete at mid...............................................
    def delete_at_mid(self,pos):
        temp=self.head
        step=1
        if pos>self.n:
            print("Your position out of Index\nlinked list contain value :",self.n,"\nyou entered",pos)
        if pos<self.n:
            print("negative indexing invalid")
        while temp is not None:
            if step==pos:
                print("temp",temp.data)
                temp.next=temp.next.next
                self.n=self.n-1
                break
            temp=temp.next
            step=step+1

    def search_value(self,data):
        temp=self.head
        while temp.next is not None:
            if temp.data==data:
                return True
        return False
    





    
    def printL(self):
        temp=self.head
        while temp is not None:
            print(temp.data)
            temp=temp.next
        pass


l=LinkedList()
l.insertatfirst(45)
l.insertatfirst(50)
l.insertatfirst(12)
l.insertatlast(1)
l.insertatlast(2)
l.insert_at_mid(4,1)
l.printL()
print("lenght:",len(l))
print("delete at first..............................")
l.delete_at_first()

l.printL()
print("lenght:",len(l))
print("delete at last......................................")
l.delete_at_last()
l.printL()
print("lenght:",len(l))
print("delete at mid...................................................")
l.delete_at_mid(-1)
l.printL()
print("lenght:",len(l))
print(l.search_value(2))
```
